```python
from sklearn.cluster import DBSCAN 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
X = np.array([[1,2],[2,2],[2,3],[8,7],[8,8],[25,80]])
```

```python
plt.scatter(X[:,0],X[:,1])
plt.show()
```

```python
db  = DBSCAN(eps=3,min_samples=2)

db.fit(X)
db.labels_
```

```python
from sklearn.datasets import make_circles 

X, _ = make_circles(n_samples=1000,factor=0.5, noise=0.03, random_state=33)
```

```python
db1 = DBSCAN(eps=0.2,min_samples=5)
clusters= db1.fit_predict(X)
```

```python
plt.scatter(X[:,0],X[:,1],c=clusters, cmap='viridis', marker='x')
plt.title('DBSCAN')
plt.show()
```
